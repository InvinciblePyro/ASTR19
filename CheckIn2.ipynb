# Unit 2 Code Demo

```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
x = np.linspace(0, 2*np.pi, 100)
y = np.sin(5 * x)

# scatter
y = y + np.random.normal(0, 0.1)
```

```python
# Plotting
plt.errorbar(
    x,
    y,
    yerr=0.1,
    fmt='o' 
)

# Limitting the y-axis
plt.ylim([-1.1, 1.1])

# label the x and y axes.
plt.xlabel("x")
plt.ylabel("y")

# save the figure as “demo.png” with a background color other than white.
plt.style.use("dark_background")
plt.savefig("demo.png")
plt.show()
```
