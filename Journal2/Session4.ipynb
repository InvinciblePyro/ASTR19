```python
import matplotlib.pyplot as plt
import numpy as np
```

```python
KingVon = np.random.vonmises(mu=0, kappa=1, size=1000)
print(KingVon[0])
```

```python
plt.hist(KingVon, 100)
plt.xlabel("Angle (radians)")
plt.ylabel("Frequency")
plt.title("Von Mises Distribution Samples")

plt.savefig("Journal2-VonMises.pdf")
plt.show()
```
